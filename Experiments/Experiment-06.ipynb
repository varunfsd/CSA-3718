```python
import pandas as pd

# Display full table columns without cutting off text in Colab
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)
pd.set_option('display.max_colwidth', None)

# Simulated Food Delivery Application Logic
def food_delivery_app(action, data):
    if action == "LOGIN":
        if len(data["phone"]) == 10 and data["phone"].isdigit() and data["otp"] == "1234":
            return "Login Successful - Redirected to Home Screen"
    elif action == "SEARCH":
        menu = ["Margherita Pizza", "Veg Biryani", "Masala Dosa"]
        if data["item"] in menu:
            return f"Item Found: {data['item']} available"
    elif action == "COUPON":
        if data["code"] == "SAVE50" and data["cart_total"] >= 200:
            return "Coupon Applied: Rs. 50 discount added"
    elif action == "ORDER":
        if data["address"] != "" and data["payment"] in ["UPI", "CARD", "COD"]:
            return "Order Confirmed - Preparing Food"
    return "Error"

# Positive Test Cases (All Valid Inputs)
test_cases = [
    {
        "TC_ID": "TC_POS_01",
        "Scenario": "Verify login with valid phone & OTP",
        "Action": "LOGIN",
        "Input_Data": {"phone": "9876543210", "otp": "1234"},
        "Expected_Result": "Login Successful - Redirected to Home Screen"
    },
    {
        "TC_ID": "TC_POS_02",
        "Scenario": "Verify searching an available food item",
        "Action": "SEARCH",
        "Input_Data": {"item": "Veg Biryani"},
        "Expected_Result": "Item Found: Veg Biryani available"
    },
    {
        "TC_ID": "TC_POS_03",
        "Scenario": "Verify applying valid promo code on eligible cart",
        "Action": "COUPON",
        "Input_Data": {"code": "SAVE50", "cart_total": 350},
        "Expected_Result": "Coupon Applied: Rs. 50 discount added"
    },
    {
        "TC_ID": "TC_POS_04",
        "Scenario": "Verify placing order with valid address & UPI",
        "Action": "ORDER",
        "Input_Data": {"address": "12 Anna Nagar, Chennai", "payment": "UPI"},
        "Expected_Result": "Order Confirmed - Preparing Food"
    }
]

# Execute Tests & Compare Expected vs Actual
results = []
for tc in test_cases:
    actual = food_delivery_app(tc["Action"], tc["Input_Data"])
    status = "PASS" if actual == tc["Expected_Result"] else "FAIL"
    results.append([tc["TC_ID"], tc["Scenario"], str(tc["Input_Data"]), tc["Expected_Result"], actual, status])

df_exp6 = pd.DataFrame(results, columns=["Test Case ID", "Test Scenario", "Test Data", "Expected Result", "Actual Result", "Status"])
print("=== EXPERIMENT 6: FOOD DELIVERY APP - POSITIVE TEST CASES ===\n")
print(df_exp6.to_string(index=False))
```
