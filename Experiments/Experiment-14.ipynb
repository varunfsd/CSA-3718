```python
%%bash
which javac >/dev/null || (apt-get update -qq && apt-get install -y -qq default-jdk)
[ -f junit.jar ] || wget -q -O junit.jar https://repo1.maven.org/maven2/org/junit/platform/junit-platform-console-standalone/1.11.4/junit-platform-console-standalone-1.11.4.jar

# 1. Java Program to Check Leap Year
cat << 'EOF' > LeapYear.java
public class LeapYear {
    public static boolean isLeapYear(int year) {
        if (year % 400 == 0) return true;
        if (year % 100 == 0) return false;
        return (year % 4 == 0);
    }
}
EOF

# 2. JUnit Test Class using assertTrue & assertFalse
cat << 'EOF' > LeapYearTest.java
import org.junit.jupiter.api.Test;
import static org.junit.jupiter.api.Assertions.assertTrue;
import static org.junit.jupiter.api.Assertions.assertFalse;

public class LeapYearTest {
    @Test
    public void testLeapYears() {
        assertTrue(LeapYear.isLeapYear(2024), "2024 is divisible by 4 -> Leap Year");
        assertTrue(LeapYear.isLeapYear(2000), "2000 is divisible by 400 -> Leap Year");
    }

    @Test
    public void testNonLeapYears() {
        assertFalse(LeapYear.isLeapYear(1900), "1900 is divisible by 100 but not 400 -> Not Leap Year");
        assertFalse(LeapYear.isLeapYear(2026), "2026 is not divisible by 4 -> Not Leap Year");
    }
}
EOF

# 3. Compile and Execute JUnit Test
javac -cp junit.jar LeapYear.java LeapYearTest.java
java -jar junit.jar execute --class-path . --select-class LeapYearTest
```
