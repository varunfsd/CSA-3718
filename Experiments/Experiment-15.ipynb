```python
%%bash
which javac >/dev/null || (apt-get update -qq && apt-get install -y -qq default-jdk)
[ -f junit.jar ] || wget -q -O junit.jar https://repo1.maven.org/maven2/org/junit/platform/junit-platform-console-standalone/1.11.4/junit-platform-console-standalone-1.11.4.jar

# 1. Java Program for Fibonacci Series
cat << 'EOF' > FibonacciBoundary.java
public class FibonacciBoundary {
    public static int getTerm(int n) {
        if (n <= 0) return 0;
        if (n == 1) return 1;
        int a = 0, b = 1;
        for (int i = 2; i <= n; i++) {
            int sum = a + b;
            a = b;
            b = sum;
        }
        return b;
    }
}
EOF

# 2. JUnit Test Class for Boundary Values (0, 1, 2)
cat << 'EOF' > FibonacciBoundaryTest.java
import org.junit.jupiter.api.Test;
import static org.junit.jupiter.api.Assertions.assertEquals;

public class FibonacciBoundaryTest {
    @Test
    public void testBoundaryTerms() {
        assertEquals(0, FibonacciBoundary.getTerm(0), "0th term boundary should be 0");
        assertEquals(1, FibonacciBoundary.getTerm(1), "1st term boundary should be 1");
        assertEquals(1, FibonacciBoundary.getTerm(2), "2nd term boundary should be 1");
    }
}
EOF

# 3. Compile and Execute JUnit Test
javac -cp junit.jar FibonacciBoundary.java FibonacciBoundaryTest.java
java -jar junit.jar execute --class-path . --select-class FibonacciBoundaryTest
```
