```python
import pandas as pd

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)
pd.set_option('display.max_colwidth', None)

# Simulated Railway Reservation Application
class RailwayReservationApp:
    def __init__(self):
        self.trains = {("Chennai", "Bengaluru"): "12607 Lalbagh Express"}
        self.seats_available = 45
        self.pnr_db = {}

    def run_action(self, action, data):
        if action == "SEARCH_TRAIN":
            route = (data["from"], data["to"])
            if route in self.trains and data["date"] != "":
                return f"Train Found: {self.trains[route]} ({self.seats_available} Seats)"
        elif action == "BOOK_TICKET":
            if data["name"] != "" and data["age"] > 0 and self.seats_available > 0:
                self.seats_available -= 1
                pnr = "4512897630"
                self.pnr_db[pnr] = "CONFIRMED - Coach S4, Berth 21"
                return f"Booking Successful - PNR: {pnr}"
        elif action == "CHECK_PNR":
            if data["pnr"] in self.pnr_db:
                return f"PNR Status: {self.pnr_db[data['pnr']]}"
        return "Operation Failed"

railway = RailwayReservationApp()

railway_test_cases = [
    {
        "TC_ID": "TC_RLY_01",
        "Scenario": "Search trains between valid source and destination",
        "Action": "SEARCH_TRAIN",
        "Input_Data": {"from": "Chennai", "to": "Bengaluru", "date": "2026-12-10"},
        "Expected_Result": "Train Found: 12607 Lalbagh Express (45 Seats)"
    },
    {
        "TC_ID": "TC_RLY_02",
        "Scenario": "Book ticket with valid passenger details",
        "Action": "BOOK_TICKET",
        "Input_Data": {"name": "Arun Kumar", "age": 24, "class": "SL"},
        "Expected_Result": "Booking Successful - PNR: 4512897630"
    },
    {
        "TC_ID": "TC_RLY_03",
        "Scenario": "Verify PNR status enquiry for booked ticket",
        "Action": "CHECK_PNR",
        "Input_Data": {"pnr": "4512897630"},
        "Expected_Result": "PNR Status: CONFIRMED - Coach S4, Berth 21"
    }
]

results = []
for tc in railway_test_cases:
    actual = railway.run_action(tc["Action"], tc["Input_Data"])
    status = "PASS" if actual == tc["Expected_Result"] else "FAIL"
    results.append([tc["TC_ID"], tc["Scenario"], str(tc["Input_Data"]), tc["Expected_Result"], actual, status])

df_exp11 = pd.DataFrame(results, columns=["Test Case ID", "Test Scenario", "Test Data", "Expected Result", "Actual Result", "Status"])
print("=== EXPERIMENT 11: RAILWAY RESERVATION SYSTEM - POSITIVE TEST CASES ===\n")
print(df_exp11.to_string(index=False))
```
