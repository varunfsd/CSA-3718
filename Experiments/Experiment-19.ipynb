```python
%%bash
which javac >/dev/null || (apt-get update -qq && apt-get install -y -qq default-jdk)
[ -f junit.jar ] || wget -q -O junit.jar https://repo1.maven.org/maven2/org/junit/platform/junit-platform-console-standalone/1.11.4/junit-platform-console-standalone-1.11.4.jar

# 1. Java Program to Calculate Compound Interest
cat << 'EOF' > CompoundInterest.java
public class CompoundInterest {
    public static double calculateCI(double principal, double rate, int years) {
        double amount = principal * Math.pow((1 + rate / 100.0), years);
        return amount - principal;
    }
}
EOF

# 2. JUnit Test Class (using 0.01 delta tolerance for double values)
cat << 'EOF' > CompoundInterestTest.java
import org.junit.jupiter.api.Test;
import static org.junit.jupiter.api.Assertions.assertEquals;

public class CompoundInterestTest {
    @Test
    public void testCompoundInterestCalculation() {
        assertEquals(210.0, CompoundInterest.calculateCI(1000.0, 10.0, 2), 0.01, "CI for 1000 at 10% for 2 yrs must be 210.0");
        assertEquals(1576.25, CompoundInterest.calculateCI(10000.0, 5.0, 3), 0.01, "CI for 10000 at 5% for 3 yrs must be 1576.25");
        assertEquals(0.0, CompoundInterest.calculateCI(5000.0, 8.0, 0), 0.01, "CI for 0 years must be 0.0");
    }
}
EOF

# 3. Compile and Execute JUnit Test
javac -cp junit.jar CompoundInterest.java CompoundInterestTest.java
java -jar junit.jar execute --class-path . --select-class CompoundInterestTest
```
