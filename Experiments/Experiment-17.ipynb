```python
%%bash
which javac >/dev/null || (apt-get update -qq && apt-get install -y -qq default-jdk)
[ -f junit.jar ] || wget -q -O junit.jar https://repo1.maven.org/maven2/org/junit/platform/junit-platform-console-standalone/1.11.4/junit-platform-console-standalone-1.11.4.jar

# 1. Java Program to Find Maximum of Three Numbers
cat << 'EOF' > MaxOfThree.java
public class MaxOfThree {
    public static int findMax(int a, int b, int c) {
        int max = a;
        if (b > max) {
            max = b;
        }
        if (c > max) {
            max = c;
        }
        return max;
    }
}
EOF

# 2. JUnit Test Class
cat << 'EOF' > MaxOfThreeTest.java
import org.junit.jupiter.api.Test;
import static org.junit.jupiter.api.Assertions.assertEquals;

public class MaxOfThreeTest {
    @Test
    public void testPositiveNumbers() {
        assertEquals(30, MaxOfThree.findMax(10, 30, 20), "Maximum of 10, 30, 20 must be 30");
        assertEquals(99, MaxOfThree.findMax(99, 45, 12), "Maximum when first number is largest");
    }

    @Test
    public void testNegativeAndEqualNumbers() {
        assertEquals(-2, MaxOfThree.findMax(-5, -2, -10), "Maximum of negative numbers must be -2");
        assertEquals(15, MaxOfThree.findMax(15, 15, 10), "Duplicate maximum values must return 15");
    }
}
EOF

# 3. Compile and Execute JUnit Test
javac -cp junit.jar MaxOfThree.java MaxOfThreeTest.java
java -jar junit.jar execute --class-path . --select-class MaxOfThreeTest
```
