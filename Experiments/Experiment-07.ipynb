```python
import pandas as pd

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)
pd.set_option('display.max_colwidth', None)

# Food Delivery App with Proper Error Messages
def food_delivery_app_validation(action, data):
    if action == "LOGIN":
        if len(data["phone"]) != 10 or not data["phone"].isdigit():
            return "Error: Enter a valid 10-digit mobile number"
        if data["otp"] != "1234":
            return "Error: Invalid OTP entered"
    elif action == "SEARCH":
        menu = ["Margherita Pizza", "Veg Biryani", "Masala Dosa"]
        if data["item"] not in menu:
            return "Error: Item currently unavailable"
    elif action == "COUPON":
        if data["code"] != "SAVE50":
            return "Error: Invalid or Expired Coupon Code"
        if data["cart_total"] < 200:
            return "Error: Minimum order value of Rs. 200 required"
    elif action == "ORDER":
        if data["address"].strip() == "":
            return "Error: Delivery address cannot be empty"
    return "Success"

# Negative Test Cases (Invalid Inputs)
neg_test_cases = [
    {
        "TC_ID": "TC_NEG_01",
        "Scenario": "Login with 6-digit invalid phone number",
        "Action": "LOGIN",
        "Input_Data": {"phone": "987654", "otp": "1234"},
        "Expected_Result": "Error: Enter a valid 10-digit mobile number"
    },
    {
        "TC_ID": "TC_NEG_02",
        "Scenario": "Login with wrong OTP",
        "Action": "LOGIN",
        "Input_Data": {"phone": "9876543210", "otp": "0000"},
        "Expected_Result": "Error: Invalid OTP entered"
    },
    {
        "TC_ID": "TC_NEG_03",
        "Scenario": "Search for a dish not on the menu",
        "Action": "SEARCH",
        "Input_Data": {"item": "Sushi Platter"},
        "Expected_Result": "Error: Item currently unavailable"
    },
    {
        "TC_ID": "TC_NEG_04",
        "Scenario": "Apply coupon when cart total is below Rs. 200",
        "Action": "COUPON",
        "Input_Data": {"code": "SAVE50", "cart_total": 120},
        "Expected_Result": "Error: Minimum order value of Rs. 200 required"
    },
    {
        "TC_ID": "TC_NEG_05",
        "Scenario": "Place order with blank delivery address",
        "Action": "ORDER",
        "Input_Data": {"address": "   "},
        "Expected_Result": "Error: Delivery address cannot be empty"
    }
]

results = []
for tc in neg_test_cases:
    actual = food_delivery_app_validation(tc["Action"], tc["Input_Data"])
    status = "PASS" if actual == tc["Expected_Result"] else "FAIL"
    results.append([tc["TC_ID"], tc["Scenario"], str(tc["Input_Data"]), tc["Expected_Result"], actual, status])

df_exp7 = pd.DataFrame(results, columns=["Test Case ID", "Test Scenario", "Test Data", "Expected Result", "Actual Result", "Status"])
print("=== EXPERIMENT 7: FOOD DELIVERY APP - NEGATIVE TEST CASES ===\n")
print(df_exp7.to_string(index=False))
```
