```python
%%bash
which javac >/dev/null || (apt-get update -qq && apt-get install -y -qq default-jdk)
[ -f junit.jar ] || wget -q -O junit.jar https://repo1.maven.org/maven2/org/junit/platform/junit-platform-console-standalone/1.11.4/junit-platform-console-standalone-1.11.4.jar

# 1. Java Program to Calculate GCD
cat << 'EOF' > GCDCalculator.java
public class GCDCalculator {
    public static int findGCD(int a, int b) {
        a = Math.abs(a);
        b = Math.abs(b);
        while (b != 0) {
            int remainder = a % b;
            a = b;
            b = remainder;
        }
        return a;
    }
}
EOF

# 2. JUnit Test Class
cat << 'EOF' > GCDCalculatorTest.java
import org.junit.jupiter.api.Test;
import static org.junit.jupiter.api.Assertions.assertEquals;

public class GCDCalculatorTest {
    @Test
    public void testGCDCalculation() {
        assertEquals(6, GCDCalculator.findGCD(12, 18), "GCD of 12 and 18 should be 6");
        assertEquals(14, GCDCalculator.findGCD(56, 98), "GCD of 56 and 98 should be 14");
        assertEquals(1, GCDCalculator.findGCD(17, 23), "GCD of two prime numbers should be 1");
    }
}
EOF

# 3. Compile and Execute JUnit Test
javac -cp junit.jar GCDCalculator.java GCDCalculatorTest.java
java -jar junit.jar execute --class-path . --select-class GCDCalculatorTest
```
