```python
%%bash
which javac >/dev/null || (apt-get update -qq && apt-get install -y -qq default-jdk)
[ -f junit.jar ] || wget -q -O junit.jar https://repo1.maven.org/maven2/org/junit/platform/junit-platform-console-standalone/1.11.4/junit-platform-console-standalone-1.11.4.jar

# 1. Java Program to Count Vowels and Consonants
cat << 'EOF' > VowelConsonantCounter.java
public class VowelConsonantCounter {
    public static int countVowels(String text) {
        if (text == null) return 0;
        int vowels = 0;
        String lower = text.toLowerCase();
        for (int i = 0; i < lower.length(); i++) {
            char ch = lower.charAt(i);
            if (ch == 'a' || ch == 'e' || ch == 'i' || ch == 'o' || ch == 'u') {
                vowels++;
            }
        }
        return vowels;
    }

    public static int countConsonants(String text) {
        if (text == null) return 0;
        int consonants = 0;
        String lower = text.toLowerCase();
        for (int i = 0; i < lower.length(); i++) {
            char ch = lower.charAt(i);
            if (Character.isLetter(ch) && !(ch == 'a' || ch == 'e' || ch == 'i' || ch == 'o' || ch == 'u')) {
                consonants++;
            }
        }
        return consonants;
    }
}
EOF

# 2. JUnit Test Class
cat << 'EOF' > VowelConsonantCounterTest.java
import org.junit.jupiter.api.Test;
import static org.junit.jupiter.api.Assertions.assertEquals;

public class VowelConsonantCounterTest {
    @Test
    public void testVowelsAndConsonants() {
        String sample = "Hello World 123";
        assertEquals(3, VowelConsonantCounter.countVowels(sample), "Vowels in 'Hello World 123' must be 3");
        assertEquals(7, VowelConsonantCounter.countConsonants(sample), "Consonants in 'Hello World 123' must be 7");
    }

    @Test
    public void testOnlyVowelsAndEmptyString() {
        assertEquals(5, VowelConsonantCounter.countVowels("AEIOU"), "AEIOU has 5 vowels");
        assertEquals(0, VowelConsonantCounter.countConsonants("AEIOU"), "AEIOU has 0 consonants");
        assertEquals(0, VowelConsonantCounter.countVowels(""), "Empty string has 0 vowels");
    }
}
EOF

# 3. Compile and Execute JUnit Test
javac -cp junit.jar VowelConsonantCounter.java VowelConsonantCounterTest.java
java -jar junit.jar execute --class-path . --select-class VowelConsonantCounterTest
```
