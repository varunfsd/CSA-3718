```python
%%bash
# Setup Java & JUnit 5 automatically (runs only once)
which javac >/dev/null || (apt-get update -qq && apt-get install -y -qq default-jdk)
[ -f junit.jar ] || wget -q -O junit.jar https://repo1.maven.org/maven2/org/junit/platform/junit-platform-console-standalone/1.11.4/junit-platform-console-standalone-1.11.4.jar

# 1. Java Program to find Nth Fibonacci Term
cat << 'EOF' > Fibonacci.java
public class Fibonacci {
    public static int findNthTerm(int n) {
        if (n <= 0) return 0;
        if (n == 1) return 1;
        int first = 0, second = 1;
        for (int i = 2; i <= n; i++) {
            int next = first + second;
            first = second;
            second = next;
        }
        return second;
    }
}
EOF

# 2. JUnit Test Class
cat << 'EOF' > FibonacciTest.java
import org.junit.jupiter.api.Test;
import static org.junit.jupiter.api.Assertions.assertEquals;

public class FibonacciTest {
    @Test
    public void testNthFibonacciTerm() {
        assertEquals(5, Fibonacci.findNthTerm(5), "5th Fibonacci term must be 5");
        assertEquals(8, Fibonacci.findNthTerm(6), "6th Fibonacci term must be 8");
        assertEquals(13, Fibonacci.findNthTerm(7), "7th Fibonacci term must be 13");
    }
}
EOF

# 3. Compile and Execute JUnit Test
javac -cp junit.jar Fibonacci.java FibonacciTest.java
java -jar junit.jar execute --class-path . --select-class FibonacciTest
```
