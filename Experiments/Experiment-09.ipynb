```python
import pandas as pd

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)
pd.set_option('display.max_colwidth', None)

def banking_negative_check(action, data):
    balance = 40000
    if action == "LOGIN":
        if data["cust_id"] != "CUST101" or data["mpin"] != "4321":
            return "Error: Invalid Customer ID or MPIN"
    elif action == "TRANSFER":
        amt = data["amount"]
        acc = data["to_account"]
        if len(acc) != 10 or not acc.isdigit():
            return "Error: Invalid 10-digit Beneficiary Account Number"
        if amt <= 0:
            return "Error: Transfer amount must be greater than Rs. 0"
        if amt > balance:
            return "Error: Insufficient Account Balance"
    return "Allowed"

neg_bank_cases = [
    {
        "TC_ID": "TC_BNK_NEG_01",
        "Scenario": "Attempt login with wrong 4-digit MPIN",
        "Action": "LOGIN",
        "Input_Data": {"cust_id": "CUST101", "mpin": "9999"},
        "Expected_Result": "Error: Invalid Customer ID or MPIN"
    },
    {
        "TC_ID": "TC_BNK_NEG_02",
        "Scenario": "Transfer amount exceeding available balance (Rs. 40000)",
        "Action": "TRANSFER",
        "Input_Data": {"to_account": "1020304050", "amount": 75000},
        "Expected_Result": "Error: Insufficient Account Balance"
    },
    {
        "TC_ID": "TC_BNK_NEG_03",
        "Scenario": "Transfer negative amount (-500)",
        "Action": "TRANSFER",
        "Input_Data": {"to_account": "1020304050", "amount": -500},
        "Expected_Result": "Error: Transfer amount must be greater than Rs. 0"
    },
    {
        "TC_ID": "TC_BNK_NEG_04",
        "Scenario": "Transfer zero amount (0)",
        "Action": "TRANSFER",
        "Input_Data": {"to_account": "1020304050", "amount": 0},
        "Expected_Result": "Error: Transfer amount must be greater than Rs. 0"
    },
    {
        "TC_ID": "TC_BNK_NEG_05",
        "Scenario": "Transfer to invalid 5-digit account number",
        "Action": "TRANSFER",
        "Input_Data": {"to_account": "12345", "amount": 2000},
        "Expected_Result": "Error: Invalid 10-digit Beneficiary Account Number"
    }
]

results = []
for tc in neg_bank_cases:
    actual = banking_negative_check(tc["Action"], tc["Input_Data"])
    status = "PASS" if actual == tc["Expected_Result"] else "FAIL"
    results.append([tc["TC_ID"], tc["Scenario"], str(tc["Input_Data"]), tc["Expected_Result"], actual, status])

df_exp9 = pd.DataFrame(results, columns=["Test Case ID", "Test Scenario", "Test Data", "Expected Result", "Actual Result", "Status"])
print("=== EXPERIMENT 9: BANKING APPLICATION - NEGATIVE TEST CASES ===\n")
print(df_exp9.to_string(index=False))
```
